```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x = np.array([0, 1, 2, 3, 4, 5, 6])
y = np.array([2.1, 3.9, 6.2, 9.5, 13.8, 19.1, 25.5])
n = len(x)

def calculate_r_squared(y_true, y_pred):
    #Calculates RSS, TSS, and R^2
    # Total Sum of Squares (TSS)
    tss = np.sum((y_true - np.mean(y_true))**2)
    # Residual Sum of Squares (RSS)
    rss = np.sum((y_true - y_pred)**2)
    # R^2
    r_squared = 1 - (rss / tss)
    return rss, tss, r_squared

#Linear Regression
print("Linear Regression Results")

# Estimate parameters and get equation (uses numpy.polyfit)
# Coefficients are [beta_1, beta_0]
coeffs_linear = np.polyfit(x, y, 1)
beta_1_linear, beta_0_linear = coeffs_linear[0], coeffs_linear[1]
print(f"Parameters: beta_0 = {beta_0_linear:.4f}, beta_1 = {beta_1_linear:.4f}")
print(f"Equation: y = {beta_0_linear:.4f} + {beta_1_linear:.4f}x")

# Compute RSS, TSS, and R^2
y_pred_linear = np.polyval(coeffs_linear, x)
rss_linear, tss, r2_linear = calculate_r_squared(y, y_pred_linear)
print(f"RSS ≈ {rss_linear:.4f}, TSS ≈ {tss:.4f}, R^2 ≈ {r2_linear:.4f}")

# Quadratic Regression
print("\nQuadratic Regression Results")

# Estimate parameters and get equation (using numpy.polyfit)
# Coefficients are [beta_2, beta_1, beta_0]
coeffs_quad = np.polyfit(x, y, 2)
beta_2_quad, beta_1_quad, beta_0_quad = coeffs_quad[0], coeffs_quad[1], coeffs_quad[2]
print(f"Parameters: beta_0 = {beta_0_quad:.4f}, beta_1 = {beta_1_quad:.4f}, beta_2 = {beta_2_quad:.4f}")
print(f"Equation: y = {beta_0_quad:.4f} + {beta_1_quad:.4f}x + {beta_2_quad:.4f}x^2")

# iii. Compute RSS, TSS, and R^2
y_pred_quad = np.polyval(coeffs_quad, x)
rss_quad, tss, r2_quad = calculate_r_squared(y, y_pred_quad)
print(f"RSS = {rss_quad:.4f}, TSS = {tss:.4f}, R^2 = {r2_quad:.4f}")

# Plotting and Model Comparison

print("\nModel Comparison")

# Create smoother x-values for plotting the curves
x_plot = np.linspace(min(x), max(x), 100)
y_plot_linear = np.polyval(coeffs_linear, x_plot)
y_plot_quad = np.polyval(coeffs_quad, x_plot)

# Plotting plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='black', label='Data Points ($x_i, y_i$)')
plt.plot(x_plot, y_plot_linear, color='blue', linestyle='--',
         label=f'Linear Fit ($R^2={r2_linear:.4f}$)')
plt.plot(x_plot, y_plot_quad, color='red', linestyle='-',
         label=f'Quadratic Fit ($R^2={r2_quad:.4f}$)')

plt.title('Comparison of Linear and Quadratic Regression Models')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

Quadratic regression is better for the dataset.
