```python
import math

def f(x, y):
    return x**2 * math.exp(-x) + y

def solve_ivp():
    # Başlangıç Değerleri
    h = 0.25
    x_end = 2.0

    # Listeler (İlk değerleri ekliyoruz: x=0, y=1)
    x_vals = [0.0]
    y_euler = [1.0]
    y_trap = [1.0]
    y_rk2 = [1.0]

    steps = int(x_end / h)

    for i in range(steps):
        xi = x_vals[-1]

        # --- 1. Euler Metodu ---
        yi_e = y_euler[-1]
        slope_e = f(xi, yi_e)
        y_next_e = yi_e + h * slope_e
        y_euler.append(y_next_e)

        # --- 2. Trapezoidal (Yamuk) Metodu ---
        # Formül: y_{n+1} = [ y_n(1 + h/2) + (h/2)(g(x_n) + g(x_{n+1})) ] / (1 - h/2)
        # Burada f(x,y) = g(x) + y formundadır. g(x) = x^2 * e^-x
        yi_t = y_trap[-1]
        g_n = xi**2 * math.exp(-xi)
        g_next = (xi + h)**2 * math.exp(-(xi + h))

        numerator = yi_t * (1 + h/2) + (h/2) * (g_n + g_next)
        denominator = 1 - h/2
        y_trap.append(numerator / denominator)

        # --- 3. RK2 (Heun) Metodu ---
        yi_r = y_rk2[-1]
        k1 = f(xi, yi_r)
        # Tahmin (Predictor)
        y_pred = yi_r + h * k1
        k2 = f(xi + h, y_pred)
        # Düzeltme (Corrector)
        y_next_r = yi_r + (h/2) * (k1 + k2)
        y_rk2.append(y_next_r)

        # x'i güncelle
        x_vals.append(xi + h)

    # Tabloyu Yazdır
    print(f"{'x':<8} {'Euler':<12} {'Trapezoidal':<12} {'RK2 (Heun)':<12}")
    print("-" * 50)
    for i in range(len(x_vals)):
        print(f"{x_vals[i]:<8.2f} {y_euler[i]:<12.4f} {y_trap[i]:<12.4f} {y_rk2[i]:<12.4f}")

if __name__ == "__main__":
    solve_ivp()
```
